# Receita por Categoria de Cliente

Receita media mensal por cliente, segmentada por categoria (baixo / medio / alto / premium).

**Motivacao:** a priorizacao de alertas de churn deve considerar receita.

**Fonte:** `cliente_mes.parquet` + `cliente_fidelidade.parquet`

```python
import sys, os, warnings
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
warnings.filterwarnings('ignore')

PROJECT_ROOT = next(p for p in [Path.cwd()] + list(Path.cwd().parents) if (p / 'src').exists())
sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

from src.config import PROCESSED_DIR, VERDE
print('Setup OK')
```

```python
cm = pd.read_parquet(PROCESSED_DIR / 'cliente_mes.parquet')
cf = pd.read_parquet(PROCESSED_DIR / 'cliente_fidelidade.parquet')
print(f'cliente_mes       : {cm.shape[0]:,} linhas  |  {cm["CLIENTE"].nunique():,} clientes')
print(f'cliente_fidelidade: {cf.shape[0]:,} clientes')
print(f'Periodo: {cm["ANO_MES"].min()} -> {cm["ANO_MES"].max()}')
```

```python
receita_cliente = (
    cm.groupby('CLIENTE')['total_valor']
    .mean()
    .reset_index()
    .rename(columns={'total_valor': 'receita_media_mensal'})
)
receita_cliente = receita_cliente.merge(
    cf[['CLIENTE', 'categoria_cliente']],
    on='CLIENTE', how='left',
)
print(f'Clientes: {len(receita_cliente):,}  | NaN categoria: {receita_cliente["categoria_cliente"].isna().sum()}')
```

```python
ORDEM = ['baixo', 'medio', 'alto', 'premium']
CORES = [VERDE[2], VERDE[3], VERDE[5], VERDE[7]]

resumo = (
    receita_cliente.groupby('categoria_cliente')['receita_media_mensal']
    .agg(
        n_clientes='count',
        media='mean',
        mediana='median',
        p25=lambda x: x.quantile(0.25),
        p75=lambda x: x.quantile(0.75),
        maximo='max',
    )
    .reindex(ORDEM)
    .round(2)
)

print('=== Receita media mensal por categoria (R$) ===\n')
print(f'{"Categoria":<10} {"Clientes":>9} {"Media":>12} {"Mediana":>12} {"P25":>12} {"P75":>12} {"Maximo":>14}')
print('-' * 85)
for cat, row in resumo.iterrows():
    print(f'{cat:<10} {int(row["n_clientes"]):>9,} {row["media"]:>12,.2f} {row["mediana"]:>12,.2f} {row["p25"]:>12,.2f} {row["p75"]:>12,.2f} {row["maximo"]:>14,.2f}')
```

```python
dados = [receita_cliente.loc[receita_cliente['categoria_cliente']==cat,'receita_media_mensal'].dropna().values for cat in ORDEM]
n_cli = [len(d) for d in dados]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

bp = axes[0].boxplot(dados, labels=ORDEM, patch_artist=True,
    flierprops=dict(marker='o', markerfacecolor='#e63946', markersize=4, alpha=0.6, linestyle='none'),
    medianprops=dict(color=VERDE[7], linewidth=2),
    whiskerprops=dict(linewidth=1.2), capprops=dict(linewidth=1.2))
for patch, cor in zip(bp['boxes'], CORES):
    patch.set_facecolor(cor); patch.set_alpha(0.8)
axes[0].set_title('Receita media mensal por categoria\n(pontos vermelhos = outliers)', fontsize=12)
axes[0].set_ylabel('Receita media mensal (R$)')
axes[0].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'R$ {x:,.0f}'))
axes[0].set_facecolor('#f4faf6')

bars = axes[1].bar(ORDEM, n_cli, color=CORES, edgecolor='white', width=0.6)
for bar, n in zip(bars, n_cli):
    axes[1].text(bar.get_x()+bar.get_width()/2, bar.get_height()+max(n_cli)*0.01, f'{n:,}', ha='center', fontsize=11)
axes[1].set_title('Numero de clientes por categoria', fontsize=12)
axes[1].set_ylabel('Clientes')
axes[1].set_facecolor('#f4faf6')

plt.suptitle('Receita e composicao por categoria', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
axes = axes.flatten()
for i, (cat, cor) in enumerate(zip(ORDEM, CORES)):
    d = receita_cliente.loc[receita_cliente['categoria_cliente']==cat,'receita_media_mensal'].dropna()
    q1, med, q3 = d.quantile([0.25, 0.50, 0.75])
    iqr = q3 - q1
    n_out = int(((d < q1-1.5*iqr)|(d > q3+1.5*iqr)).sum())
    axes[i].boxplot(d.values, patch_artist=True,
        boxprops=dict(facecolor=cor, alpha=0.8),
        medianprops=dict(color=VERDE[7], linewidth=2.5),
        flierprops=dict(marker='o', markerfacecolor='#e63946', markersize=5, alpha=0.7, linestyle='none'),
        whiskerprops=dict(linewidth=1.2), capprops=dict(linewidth=1.2))
    axes[i].set_title(f'{cat.capitalize()}  -  {len(d):,} clientes  |  {n_out} outliers\nMediana R$ {med:,.0f}   P25 R$ {q1:,.0f}   P75 R$ {q3:,.0f}', fontsize=11)
    axes[i].set_ylabel('Receita media mensal (R$)')
    axes[i].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'R$ {x:,.0f}'))
    axes[i].set_xticks([])
    axes[i].set_facecolor('#f4faf6')
plt.suptitle('Receita media mensal por categoria - detalhe individual', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
x = np.arange(len(ORDEM)); w = 0.35
bm = ax.bar(x-w/2, resumo['media'],   width=w, color=VERDE[5], label='Media',   edgecolor='white')
bd = ax.bar(x+w/2, resumo['mediana'], width=w, color=VERDE[3], label='Mediana', edgecolor='white')
for bar in list(bm)+list(bd):
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+resumo['media'].max()*0.01,
            f'R$ {bar.get_height():,.0f}', ha='center', va='bottom', fontsize=8.5)
ax.set_xticks(x); ax.set_xticklabels(ORDEM)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'R$ {v:,.0f}'))
ax.set_title('Media vs Mediana por categoria\n(diferenca grande = outliers puxando a media)', fontsize=12)
ax.set_ylabel('Receita media mensal (R$)')
ax.legend(); ax.set_facecolor('#f4faf6')
plt.tight_layout(); plt.show()

print('\nRazao media/mediana (> 1.5 = outliers distorcendo):')
for cat in ORDEM:
    r = resumo.loc[cat,'media'] / resumo.loc[cat,'mediana']
    print(f'  {cat:<10}: {r:.2f}{"  <- distorcao relevante" if r > 1.5 else ""}')
```
